```python

import pandas as pd
from sklearn.naive_bayes import CategoricalNB

# 1. Load Dataset
df = pd.read_csv("heart.csv")

# 2. Extract Features
X = df[["age", "sex", "cp", "chol"]].copy()

# Convert continuous values into categories
X["age"] = (X["age"] > 50).astype(int)
X["chol"] = (X["chol"] > 240).astype(int)

# Target variable
y = (df["target"] > 0).astype(int)

# 3. Train Model
model = CategoricalNB()
model.fit(X, y)

# 4. Predict for Sample Patient
patient = pd.DataFrame(
    [[1, 1, 3, 1]],
    columns=["age", "sex", "cp", "chol"]
)

probs = model.predict_proba(patient)[0]

# 5. Display Output
print("No Disease Prob :", round(probs[0], 4))
print("Disease Prob    :", round(probs[1], 4))

if model.predict(patient)[0] == 1:
    print("Diagnosis       : Heart Disease")
else:
    print("Diagnosis       : Normal")
```
